````python
from youtube_transcript_api import YouTubeTranscriptApi, TranscriptsDisabled, NoTranscriptFound

video_id = "Gfr50f6ZBvo"

try:
    transcript_list = YouTubeTranscriptApi().fetch(video_id,languages=["en"])
    transcript = " ".join(chunk.text for chunk in transcript_list)
    print(transcript)
except TranscriptsDisabled:
    print("No captions available for this video.")
except NoTranscriptFound:
    print("No transcript found for this video.")
````

````python
# 1b-Indexing:text splitting
splitter=RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=200)
chunks=splitter.create_documents([transcript])
````

--why chunks=splitter.create_documents([transcript]) and chunks=splitter.create_documents(transcript) are different?
**Because `create_documents` expects a list of strings, not a single string.**

---

### Case 1: `splitter.create_documents([transcript])` ✅ Correct

- You pass a **list containing one string**
- It treats the entire transcript as **one document**
- Then splits it into chunks
- Result: chunks from one document

### Case 2: `splitter.create_documents(transcript)` ❌ Wrong

- You pass a **single string**
- It iterates over each **character** (since a string is iterable)
- Each character becomes a separate "document"
- Result: 1,000s of chunks, each 1 character long — garbage

---

### Example

```python
transcript = "Hello world"

splitter.create_documents([transcript])
# → treats "Hello world" as ONE document → splits into chunks

splitter.create_documents(transcript)
# → treats "H", "e", "l", "l", "o", " ", "w"... as separate documents
```

---

### Bottom Line

| Code | What Happens |
|------|-------------|
| `[transcript]` | ✅ One document, split correctly |
| `transcript` | ❌ Each character = one document |


````python
# 1c and 1d-Indexing(Embedding Generation and Storing in Vector store)
embeddings=HuggingFaceEmbeddings(model="sentence-transformers/all-MiniLM-L6-v2")
vector_store=FAISS.from_documents(chunks,embeddings)
print(vector_store.index_to_docstore_id)
vector_store.save_local("faiss_index")
````

--Am i running Huggingfaceembeddings model locally?

What happens:

First run: It downloads the model weights (~80 MB) from Hugging Face Hub to your local cache folder

Location: C:\Users\admin\.cache\huggingface\

Then: It loads the model into your own RAM

Every embedding after that: Computed on your CPU — no internet needed

````python
vector_store.get_by_ids(['38cea541-f9c9-437b-90a9-14509f649c89'])
````

````python
# Retriever
retriever=vector_store.as_retriever(search_type="similarity",search_kwargs={"k":4})
````

````python
retriever
````

````python
retriever.invoke("What is deepmind")
````

````python
# Step3 -Augmentation
llm=HuggingFaceEndpoint(
    model="meta-llama/Llama-3.1-8B-Instruct",task="text-generation",max_new_tokens=35
)
model=ChatHuggingFace(llm=llm)
````

````python
prompt=PromptTemplate(
    template="""
    You are helpful assistant.Answer only from the provided transcript.If context is insufficient say I dont know.
    
    {context}
    Question:{question}
    """,
    input_variables=["context","question"]
    
)
````

````python
question="is topic of aliens discussed in the video?if yes what was discussed?"
retrieved_docs=retriever.invoke(question)
````

````python
retrieved_docs
````

````python
context_text="\n\n".join(doc.page_content for doc in retrieved_docs)
````

````python
context_text
````

````python
final_prompt=prompt.invoke({"context":context_text,"question":question})
````

````python
final_prompt
````

````python
# 4.Generation
answer=model.invoke(final_prompt)
print(answer)
print(answer.content)
````

````python
# Building a chain
from langchain_core.runnables import RunnablePassthrough,RunnableLambda,RunnableParallel
from langchain_core.output_parsers import StrOutputParser
````

````python
def format_data(retrieved_docs):
    context_text="\n\n".join(doc.page_content for doc in retrieved_docs)
    return context_text
````

````python
parallel_chain=RunnableParallel({
    "context":retriever | RunnableLambda(format_data),
    "question":RunnablePassthrough()
})
````

````python
parallel_chain.invoke("who is demis")
````

````python
parser=StrOutputParser()
````

````python
main_chain=parallel_chain | prompt | model | parser
````

````python
main_chain.invoke("Can you summarize video in 1 line")
````

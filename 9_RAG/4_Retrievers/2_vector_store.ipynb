```python
from langchain_core.documents import Document
from langchain_community.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
# Step1-Your source Document
document=[
    Document(page_content="LangChain – A framework for building applications that use language models for reasoning and data interaction.",id="Langchain"),
    Document(page_content="Chroma – An open-source vector database for storing and querying embeddings efficiently",id="Chroma"),
    Document(page_content="Embeddings – Numerical vector representations of text or other data that capture semantic meaning.",id="Embeddings"),
]
```

```python
# Iniitialize embedding model
embedding_model=HuggingFaceEmbeddings(
    model="sentence-transformers/all-MiniLM-L6-v2",
)
```

```python
# Create Chroma vector store in memory
vector_store=Chroma(
    persist_directory="my_chroma_data",
    embedding_function=embedding_model,
    collection_name="artificial_intelligence"
)
```

```python
# or we could have wrote like this

# vector_store=Chroma.from_documents(
#     documents=document,
#     embedding=embedding_model,
#     collection_name="my_collection"
# )

vector_store.add_documents(document)
```

```python
# Step4: Convert vector store into retriever
retriever=vector_store.as_retriever(search_kwargs={"k":2})
```

```python
query="What is Chroma used for?"
results=retriever.invoke(query)
```

```python
for i,doc in enumerate(results):
    print(f"\n ---Result{i+1}---")
    print(doc.page_content)
```

```python
results=vector_store.similarity_search(query,k=2)
results
```
